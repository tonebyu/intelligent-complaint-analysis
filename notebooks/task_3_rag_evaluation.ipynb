```python
# notebooks/task_3_rag_evaluation.ipynb

import sys, os
sys.path.append(os.path.abspath(".."))
```

```python
import pandas as pd
from src.rag_pipeline import load_vector_store, get_llm, answer_question
```

```python
# Load vector DB and model
vectordb, metadata, embedder = load_vector_store("../vector_store")
llm = get_llm()
```

```python
# Define test questions
questions = [
    "Why are people unhappy with Buy Now, Pay Later?",
    "What issues do users report about personal loans?",
    "Are there complaints about savings account accessibility?",
    "What do users say about money transfer delays?",
    "Do credit card users face billing errors?",
    "How often do people complain about interest rates?",
    "What are the most common issues with CrediTrust credit cards?",
    "Do customers report fraud in BNPL transactions?",
    "Are there complaints about customer service for savings accounts?",
    "How is the user experience with the mobile app?"
]
```

```python

# Collect results
results = []

for question in questions:
    answer, docs = answer_question(question, vectordb, embedder, llm, k=5)
    sources = [doc.page_content[:200].replace("\n", " ") for doc in docs[:2]]
    results.append({
        "Question": question,
        "Generated Answer": answer,
        "Source 1": sources[0],
        "Source 2": sources[1],
        "Quality Score (1–5)": "",
        "Comments": ""
    })
```

```python
# Convert to DataFrame for export or Markdown
df_results = pd.DataFrame(results)
df_results.to_markdown("../reports/task_3_rag_eval_table.md", index=False)
print("Evaluation table saved to reports/task_3_rag_eval_table.md")
```
