```python
#  1: Setup
import sys, os
sys.path.append(os.path.abspath(".."))
```

```python
from src.embedding import ComplaintChunker, TextEmbedder, create_faiss_index, save_index
import pandas as pd
```

```python
#  Load cleaned data
df = pd.read_csv("../data/filtered_complaints.csv")
texts = df["cleaned_narrative"].tolist()
```

```python
#  Chunk text
chunker = ComplaintChunker(chunk_size=200, chunk_overlap=20)
chunks = chunker.chunk(texts)
chunk_texts = [doc.page_content for doc in chunks]
```

```python
# Embed chunks
embedder = TextEmbedder()
embeddings = embedder.embed_texts(chunk_texts)
```

```python
# Create FAISS index
import numpy as np
index = create_faiss_index(np.array(embeddings))
```

```python
#  Save index + metadata
metadata = [{"original_text": t} for t in chunk_texts]
save_index(index, metadata)

print("✅ Task 2 completed: FAISS index and metadata saved to vector_store/")
```
